```python
import numpy as np

# 1. Dataset Preparation (House Area vs Price)

# x_data: Area in 100 sq meters (e.g., 1.0 = 100 sqm)
# y_data: Price in arbitrary units (e.g., billions)
x_data = np.array([0.5, 0.8, 1.1, 1.5, 2.0]) 
y_data = np.array([1.2, 1.9, 2.4, 3.5, 4.2])
```

```python
# 2. Define the Cost Function (MSE)

def compute_mse_cost(x, y, w, b):
    """
    Computes the Mean Squared Error (MSE) cost.
    Formula: J(w,b) = (1/N) * sum( (y_i - y_hat_i)^2 )
    """
    n = len(x)
    predictions = w * x + b
    squared_errors = (y - predictions) ** 2
    cost = (1 / n) * np.sum(squared_errors)
    return cost
```

```python
# 3. Define Gradients Calculation

def compute_gradients(x, y, w, b):
    """
    Computes the partial derivatives (gradients) of MSE 
    with respect to parameters w and b.
    """
    n = len(x)
    predictions = w * x + b
    
    # Derivative w.r.t w: dJ/dw = (2/N) * sum(-x * (y - y_hat))
    # which is equivalent to (2/N) * sum(x * (y_hat - y))
    dj_dw = (2 / n) * np.sum((predictions - y) * x)
    
    # Derivative w.r.t b: dJ/db = (2/N) * sum(y_hat - y)
    dj_db = (2 / n) * np.sum(predictions - y)
    
    return dj_dw, dj_db
```

```python
# 4. Gradient Descent Optimization

def gradient_descent(x, y, starting_w, starting_b, learning_rate, epochs):
    """
    Performs the Gradient Descent algorithm to optimize w and b.
    """
    w = starting_w
    b = starting_b
    cost_history = []

    for i in range(epochs):
        # 1. Calculate the gradients at current w, b
        dj_dw, dj_db = compute_gradients(x, y, w, b)
        
        # 2. Update parameters using the learning rate (eta)
        w = w - learning_rate * dj_dw
        b = b - learning_rate * dj_db
        
        # 3. Track the cost to monitor convergence/divergence
        current_cost = compute_mse_cost(x, y, w, b)
        cost_history.append(current_cost)
        
    return w, b, cost_history
```

```python
# 5. Experiments: Testing Different Learning Rates (eta)

initial_w, initial_b = 0.0, 0.0
total_epochs = 50

# Experiment A: Optimal Learning Rate
eta_optimal = 0.1
w_opt, b_opt, cost_opt = gradient_descent(x_data, y_data, initial_w, initial_b, eta_optimal, total_epochs)
print(f"Optimal eta ({eta_optimal}): Final Cost = {cost_opt[-1]:.4f}")
print(f"Physical meaning -> Base Price (b) = {b_opt:.2f}, Price per unit area (w) = {w_opt:.2f}\n")

# Experiment B: Learning Rate TOO LARGE 
# Notice we only run a few epochs because it diverges quickly
eta_large = 1.15 
w_div, b_div, cost_div = gradient_descent(x_data, y_data, initial_w, initial_b, eta_large, 5)
print(f"Large eta ({eta_large}): Cost after 5 epochs = {cost_div[-1]:.2e} (Diverging/Exploding!)")

# Experiment C: Learning Rate TOO SMALL 
eta_small = 0.0001
w_slow, b_slow, cost_slow = gradient_descent(x_data, y_data, initial_w, initial_b, eta_small, total_epochs)
print(f"Small eta ({eta_small}): Cost after {total_epochs} epochs = {cost_slow[-1]:.4f} (Converging extremely slowly)")
```
