```python
import numpy as np
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Perceptron Concept & Decision Making

def step_activation(z):
    """
    Step activation function for a basic Perceptron.
    Returns 1 if z >= 0 (Class 1), else 0 (Class 0).
    """
    return 1 if z >= 0 else 0

def perceptron_predict(x, w, b):
    """
    Predicts the class of input x based on z = wx + b
    """
    z = np.dot(w, x) + b
    return step_activation(z)

# Example: Weights and bias for a simple 2D feature space
w_example = np.array([0.5, -0.8])
b_example = 0.2

# Test sample 1
x1 = np.array([1.0, 0.5])
z1 = np.dot(w_example, x1) + b_example # z = 0.3
print(f"Sample 1: z = {z1:.2f} -> Classified as: {perceptron_predict(x1, w_example, b_example)}")

# Test sample 2
x2 = np.array([1.0, 1.5])
z2 = np.dot(w_example, x2) + b_example # z = -0.5
print(f"Sample 2: z = {z2:.2f} -> Classified as: {perceptron_predict(x2, w_example, b_example)}")
```

```python
# 2. Confusion Matrix Simulation (Spam Detection)

# 1 = Spam (Positive), 0 = Not Spam / Important Email (Negative)
# True labels of emails received
y_true = np.array([0, 0, 1, 0, 1, 1, 0, 0, 1, 0])

# Model's predictions
# Notice: index 1 was 0(Not Spam) but predicted 1(Spam) -> False Positive
# Notice: index 4 was 1(Spam) but predicted 0(Not Spam) -> False Negative
y_pred = np.array([0, 1, 1, 0, 0, 1, 0, 0, 1, 0])

# Generate Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
tn, fp, fn, tp = cm.ravel()

print("\n--- Spam Detection Confusion Matrix ---")
print(f"True Negatives (Correctly identified normal emails): {tn}")
print(f"False Positives (IMPORTANT EMAIL LOST IN SPAM!): {fp}")
print(f"False Negatives (Spam reached Inbox): {fn}")
print(f"True Positives (Correctly caught spam): {tp}")
```
